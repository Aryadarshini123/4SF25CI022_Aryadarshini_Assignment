```python
import pandas as pd
import numpy as np
```

1. Create a DataFrame and add a product column

```python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']

print(df)
```

2. Calculate row-wise and column-wise sum

```python
df = pd.DataFrame(np.random.randint(1, 20, (4, 3)),
                  columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

3. Introduce NaN values and fill them with column mean

```python
df = pd.DataFrame(np.random.randint(1, 20, (5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Before filling:")
print(df)

df = df.fillna(df.mean())

print("After filling:")
print(df)
```

4. Drop rows containing NaN values

```python
df = pd.DataFrame(np.random.randint(1, 20, (6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan

print("Before dropping:")
print(df)

df = df.dropna()

print("After dropping:")
print(df)
```

5. Group by Category and calculate sum and mean

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print(result)
```

6.Calculate total sales for each category

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Box'],
    'Category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'Sales': np.random.randint(100, 1000, 6)
})

print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total sales:")
print(result)
```

7. Merge two DataFrames using a common column

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Arya', 'Anu', 'Riya']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [90, 85, 95]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```

8. Create a MultiIndex DataFrame and perform indexing and slicing

```python
arrays = [
    ['A', 'A', 'B', 'B', 'C', 'C'],
    [1, 2, 1, 2, 1, 2]
]

index = pd.MultiIndex.from_arrays(
    arrays, names=['Category', 'Number']
)

df = pd.DataFrame(
    {'Value': [10, 20, 30, 40, 50, 60]},
    index=index
)

print("Original DataFrame:")
print(df)

print("Access Category A:")
print(df.loc['A'])

print("Access Category B:")
print(df.loc['B'])

print("Slicing:")
print(df.loc[('A', 1):('B', 2)])
```

9. MultiIndex with Category and SubCategory

```python
categories = ['A', 'A', 'B', 'B', 'C', 'C']
subcategories = ['X', 'Y', 'X', 'Y', 'X', 'Y']

index = pd.MultiIndex.from_arrays(
    [categories, subcategories],
    names=['Category', 'SubCategory']
)

df = pd.DataFrame(
    {'Value': np.random.randint(10, 100, 6)},
    index=index
)

print(df)

print("Sum by Category and SubCategory:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

10. Apply a function to double all DataFrame values

```python
df = pd.DataFrame(
    np.random.randint(1, 20, (5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

result = df.map(lambda x: x * 2)

print("After doubling:")
print(result)
```

11. Apply lambda function to calculate row-wise sum

```python
df = pd.DataFrame(
    np.random.randint(1, 20, (6, 3)),
    columns=['A', 'B', 'C']
)

df['Total'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)

print(df)
```

12. Convert Pandas Series strings to uppercase

```python
s = pd.Series([
    'apple',
    'banana',
    'mango',
    'orange',
    'grapes'
])

print("Original Series:")
print(s)

print("Uppercase Series:")
print(s.str.upper())
```

13.Extract the first three characters of each string

```python
s = pd.Series([
    'apple',
    'banana',
    'mango',
    'orange',
    'grapes'
])

print("Original Series:")
print(s)

print("First three characters:")
print(s.str[:3])
```
